**Description:** This program will allow a user to enter the last name, first name, and GPA of a student to determine if they qualify for the Dean's List or the Honor Roll. 

```python
last_name = input("Enter the student's last name: (To exit this program, enter 'zzz')")

while last_name != "zzz" and last_name != "ZZZ":
    first_name = input("Enter the student's first name:")
    gpa = input("Enter the student's GPA:")
    student_gpa = float(gpa)

    if student_gpa >= 3.5:
        print(f"{first_name} {last_name} has made the Dean's List.")

    elif student_gpa < 3.5 and student_gpa >= 3.25:
        print(f"{first_name} {last_name} has made the Honor Roll.")

    last_name = input("Enter the student's last name: (To exit this program, enter 'zzz')")
```
