```python
import sqlite3

conn = sqlite3.connect('books.db')
curs = conn.cursor()
curs.execute('''CREATE TABLE books
    (title TEXT, author TEXT, year INT)''')
conn.commit()
conn.close()
```

```python
import csv
import sqlite3

conn = sqlite3.connect('books.db')
curs = conn.cursor()

with open ('books2.csv', 'rt') as fin:
    reader = csv.DictReader(fin)
    for row in reader:
        curs.execute('INSERT INTO books (title, author, year) VALUES (?, ?, ?)',
                     (row['title'], row['author'], int(row['year'])))

conn.commit()
conn.close()
                     
```
